```python
import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from PIL import Image
from google.colab import files
from IPython.display import display

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)
```

```python
uploaded = files.upload()

if not uploaded:
    raise ValueError("Please upload a leaf image.")

image_path = next(iter(uploaded))

image = Image.open(image_path).convert("RGB")

print("Image:", image_path)
print("Size:", image.size)

display(image)
```

```python
import torchvision
from torchvision import models, transforms

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

cnn = models.resnet18(
    weights=models.ResNet18_Weights.DEFAULT
)

# Remove ImageNet classifier
cnn.fc = nn.Identity()

cnn = cnn.to(device)
cnn.eval()

image_tensor = transform(image).unsqueeze(0).to(device)

with torch.no_grad():
    cnn_features = cnn(image_tensor)

print("CNN features:", cnn_features.shape)
```

```python
from transformers import AutoTokenizer, AutoModel

MODEL_NAME = "distilbert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
transformer = AutoModel.from_pretrained(MODEL_NAME)

transformer = transformer.to(device)
transformer.eval()

farmer_text = input(
    "Describe the problem with the crop: "
)

inputs = tokenizer(
    farmer_text,
    return_tensors="pt",
    truncation=True,
    padding=True,
    max_length=128
)

inputs = {
    k: v.to(device)
    for k, v in inputs.items()
}

with torch.no_grad():
    outputs = transformer(**inputs)

text_features = outputs.last_hidden_state.mean(dim=1)

print("Transformer features:", text_features.shape)
```

```python
class DiseaseLSTM(nn.Module):

    def __init__(self, input_size=3, hidden_size=128):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True
        )

        self.fc = nn.Linear(hidden_size, 128)

    def forward(self, x):

        output, (hidden, cell) = self.lstm(x)

        return self.fc(hidden[-1])


lstm = DiseaseLSTM().to(device)

lstm.eval()
```

```python
print("Enter crop history")
print("Use values from 0 to 100.")

n = int(input("Number of observations: "))

history = []

for i in range(n):

    print(f"\nObservation {i+1}")

    affected = float(
        input("Affected area (%): ")
    )

    yellowing = float(
        input("Yellowing severity (0-10): ")
    )

    spots = float(
        input("Spot severity (0-10): ")
    )

    history.append([
        affected / 100,
        yellowing / 10,
        spots / 10
    ])

history = np.array(
    history,
    dtype=np.float32
)

history_tensor = torch.tensor(
    history,
    dtype=torch.float32
).unsqueeze(0).to(device)

with torch.no_grad():

    lstm_features = lstm(
        history_tensor
    )

print("LSTM features:", lstm_features.shape)
```

```python
class MultimodalFusion(nn.Module):

    def __init__(self):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(512 + 128 + 768, 512),

            nn.ReLU(),

            nn.Dropout(0.3),

            nn.Linear(512, 256),

            nn.ReLU(),

            nn.Linear(256, 128),

            nn.ReLU()
        )

    def forward(
        self,
        image_features,
        history_features,
        text_features
    ):

        combined = torch.cat(
            [
                image_features,
                history_features,
                text_features
            ],
            dim=1
        )

        return self.network(combined)


fusion_model = MultimodalFusion().to(device)

fusion_model.eval()

with torch.no_grad():

    fused_features = fusion_model(
        cnn_features,
        lstm_features,
        text_features
    )

print("Fused features:", fused_features.shape)
```

```python
knowledge_base = [

    {
        "title": "Tomato Early Blight",
        "content": """
        Early blight can cause brown or dark spots on tomato leaves.
        Older leaves may be affected first. Good sanitation,
        appropriate plant spacing and reducing prolonged leaf
        wetness can help manage disease development.
        """
    },

    {
        "title": "Tomato Late Blight",
        "content": """
        Late blight may produce dark irregular lesions on tomato
        leaves and stems. Disease development can be rapid under
        favorable environmental conditions. Monitoring and removing
        infected plant material are important management practices.
        """
    },

    {
        "title": "Powdery Mildew",
        "content": """
        Powdery mildew commonly produces white powder-like growth
        on leaf surfaces. Good air circulation and environmental
        management can help reduce disease development.
        """
    },

    {
        "title": "Leaf Spot",
        "content": """
        Leaf spot diseases may produce circular or irregular spots
        on leaves. Symptoms vary according to crop and pathogen.
        Good sanitation and removal of severely affected material
        may help reduce spread.
        """
    },

    {
        "title": "Nutrient Deficiency",
        "content": """
        Yellow leaves can sometimes be associated with nutrient
        deficiencies rather than infectious diseases. Leaf age,
        yellowing pattern, soil conditions and crop history should
        be considered.
        """
    }
]

print("Knowledge documents:", len(knowledge_base))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    item["title"] + " " + item["content"]
    for item in knowledge_base
]

vectorizer = TfidfVectorizer(
    stop_words="english"
)

document_vectors = vectorizer.fit_transform(
    documents
)


def retrieve_knowledge(query, top_k=3):

    query_vector = vectorizer.transform([query])

    similarity = cosine_similarity(
        query_vector,
        document_vectors
    )[0]

    indices = similarity.argsort()[-top_k:][::-1]

    results = []

    for index in indices:

        results.append({
            "title": knowledge_base[index]["title"],
            "content": knowledge_base[index]["content"],
            "score": float(similarity[index])
        })

    return results
```

```python
results = retrieve_knowledge(
    farmer_text,
    top_k=3
)

print("\n===== RAG RESULTS =====\n")

for result in results:

    print("SOURCE:", result["title"])
    print("SCORE:", round(result["score"], 3))
    print(result["content"].strip())
    print("-" * 60)
```

```python
def progression_analysis(history):

    affected = history[:, 0] * 100

    if len(affected) < 2:
        return "Not enough historical observations."

    change = affected[-1] - affected[0]

    if change > 30:
        return "Rapid increase in affected area."

    elif change > 10:
        return "Moderate increase in affected area."

    elif change > 0:
        return "Small increase in affected area."

    elif change == 0:
        return "Affected area appears stable."

    else:
        return "Affected area has decreased."


progression = progression_analysis(history)

print("Disease progression:")
print(progression)
```

```python
def prototype_assessment(text, rag_results):

    text = text.lower()

    if "white" in text and "powder" in text:
        return "Possible powdery mildew pattern"

    if "brown" in text and "spot" in text:
        return "Possible leaf spot / early blight pattern"

    if "dark" in text and "lesion" in text:
        return "Possible leaf disease pattern"

    if "yellow" in text and "spot" in text:
        return "Possible leaf spot / stress pattern"

    if "yellow" in text:
        return "Possible nutrient stress or disease pattern"

    return "Insufficient evidence for a specific assessment"


assessment = prototype_assessment(
    farmer_text,
    results
)

print(assessment)
```
