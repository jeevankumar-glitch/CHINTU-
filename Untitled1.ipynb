```python
import os
import shutil
import glob

from google.colab import files

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_huggingface import HuggingFaceEmbeddings

from langchain_community.vectorstores import FAISS

from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    GoogleGenerativeAIEmbeddings
)

from langchain_core.prompts import ChatPromptTemplate

print("All libraries loaded successfully!")
```

```python
import getpass

GOOGLE_API_KEY = getpass.getpass(
    "Enter your Google Gemini API Key: "
)

os.environ["GOOGLE_API_KEY"] = GOOGLE_API_KEY

print("API key configured.")
```

```python
print("Upload your college PDF documents.")

uploaded_files = files.upload()

pdf_files = list(uploaded_files.keys())

print("\nUploaded files:")

for file in pdf_files:
    print("-", file)

if len(pdf_files) == 0:
    raise ValueError("No PDF files uploaded.")
```

```python
documents = []

for pdf in pdf_files:

    print(f"\nLoading: {pdf}")

    loader = PyPDFLoader(pdf)

    docs = loader.load()

    documents.extend(docs)

print("\nTotal pages loaded:", len(documents))
```

```python
for doc in documents:

    source = doc.metadata.get("source", "Unknown")

    page = doc.metadata.get("page", 0)

    doc.metadata["source_file"] = os.path.basename(source)

    doc.metadata["page_number"] = page + 1


print("Metadata added successfully.")
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

chunks = text_splitter.split_documents(
    documents
)

print("Original pages:", len(documents))
print("Text chunks:", len(chunks))
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)

print("Embedding model loaded.")
```

```python
print("Creating vector database...")

vector_db = FAISS.from_documents(
    chunks,
    embeddings
)

print("FAISS vector database created successfully.")
```

```python
retriever = vector_db.as_retriever(
    search_type="similarity",
    search_kwargs={
        "k": 5
    }
)

print("Retriever ready.")
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.2
)

print("Gemini model loaded.")
```

```python
prompt = ChatPromptTemplate.from_template("""

You are CampusGPT, an intelligent college knowledge assistant.

Your job is to answer student questions using ONLY the information
provided in the retrieved college documents.

Rules:

1. Do not invent information.
2. Do not use outside knowledge when answering document-specific questions.
3. If the answer cannot be found in the documents, clearly say:
   "I could not find this information in the uploaded college documents."
4. Give a concise but useful answer.
5. Include the source document and page number whenever possible.
6. If multiple documents contain relevant information, combine them carefully.
7. Distinguish between directly stated information and reasonable interpretation.
8. Never pretend that information exists in a document when it does not.

Retrieved Context:

{context}

Student Question:

{question}

Answer:
""")
```

```python
def format_documents(docs):

    formatted = []

    for doc in docs:

        source = doc.metadata.get(
            "source_file",
            "Unknown"
        )

        page = doc.metadata.get(
            "page_number",
            "Unknown"
        )

        text = doc.page_content

        formatted.append(
            f"""
SOURCE: {source}
PAGE: {page}

{text}
"""
        )

    return "\n\n".join(formatted)
```

```python
def campus_gpt(question):

    # Retrieve relevant documents
    docs = retriever.invoke(question)

    # Format retrieved context
    context = format_documents(docs)

    # Create prompt
    messages = prompt.invoke({
        "context": context,
        "question": question
    })

    # Ask Gemini
    response = llm.invoke(messages)

    answer = response.content

    # Collect sources
    sources = []

    for doc in docs:

        source = doc.metadata.get(
            "source_file",
            "Unknown"
        )

        page = doc.metadata.get(
            "page_number",
            "Unknown"
        )

        item = f"{source} — Page {page}"

        if item not in sources:
            sources.append(item)

    return answer, sources
```

```python
question = input(
    "Ask CampusGPT a question: "
)

answer, sources = campus_gpt(
    question
)

print("\n" + "=" * 70)

print("CAMPUSGPT")
print("=" * 70)

print("\nANSWER:\n")
print(answer)

print("\nSOURCES:\n")

for source in sources:
    print("📄", source)
```

```python
chat_history = []

def campus_chat(question):

    docs = retriever.invoke(question)

    context = format_documents(docs)

    history_text = ""

    for user_msg, bot_msg in chat_history[-5:]:

        history_text += f"""
Student: {user_msg}
CampusGPT: {bot_msg}

"""

    chat_prompt = ChatPromptTemplate.from_template("""

You are CampusGPT.

Answer the student's question using the retrieved college
documents.

Conversation history:

{history}

Retrieved documents:

{context}

Current question:

{question}

Instructions:

- Use the retrieved documents as the primary source.
- Do not invent college-specific information.
- If the answer is unavailable, say so.
- Keep the answer clear and student-friendly.
- Include document/page sources.

Answer:
""")

    messages = chat_prompt.invoke({
        "history": history_text,
        "context": context,
        "question": question
    })

    response = llm.invoke(messages)

    answer = response.content

    chat_history.append(
        (question, answer)
    )

    return answer, docs
```

```python
print("CampusGPT is ready!")
print("Type 'exit' to stop.\n")

while True:

    question = input("You: ")

    if question.lower() == "exit":
        print("CampusGPT: Goodbye!")
        break

    answer, docs = campus_chat(
        question
    )

    print("\nCampusGPT:")
    print(answer)

    print("\nSources:")

    shown = set()

    for doc in docs:

        source = doc.metadata.get(
            "source_file",
            "Unknown"
        )

        page = doc.metadata.get(
            "page_number",
            "Unknown"
        )

        source_text = f"{source} — Page {page}"

        if source_text not in shown:

            print("📄", source_text)

            shown.add(source_text)

    print("\n" + "-" * 70 + "\n")
```

```python
def summarize_document(document_text):

    summary_prompt = ChatPromptTemplate.from_template("""

Summarize the following college document.

Provide:

1. Document purpose
2. Important dates
3. Important rules
4. Important academic information
5. Important requirements
6. Important points students should know

Document:

{text}

Do not invent information.

""")

    messages = summary_prompt.invoke({
        "text": document_text[:30000]
    })

    response = llm.invoke(messages)

    return response.content
```

```python
full_text = "\n\n".join(
    doc.page_content
    for doc in documents
)

summary = summarize_document(
    full_text
)

print(summary)
```

```python
def generate_questions(subject):

    docs = retriever.invoke(
        f"important topics and questions for {subject}"
    )

    context = format_documents(docs)

    question_prompt = ChatPromptTemplate.from_template("""

You are an academic assistant.

Using ONLY the retrieved college material, generate:

- 5 important long-answer questions
- 5 short-answer questions
- 5 conceptual questions

Subject:

{subject}

Retrieved material:

{context}

Do not invent topics that are unrelated to the retrieved material.

""")

    messages = question_prompt.invoke({
        "subject": subject,
        "context": context
    })

    response = llm.invoke(messages)

    return response.content
```

```python
subject = input(
    "Enter subject name: "
)

questions = generate_questions(
    subject
)

print(questions)
```

```python
import gradio as gr

def gradio_chat(message, history):

    answer, sources = campus_gpt(
        message
    )

    source_text = "\n\n**Sources:**\n"

    for source in sources:
        source_text += f"- {source}\n"

    return answer + source_text


demo = gr.ChatInterface(
    fn=gradio_chat,
    title="🎓 CampusGPT",
    description=(
        "LangChain-Based Intelligent College Knowledge Assistant"
    ),
    examples=[
        "What are the eligibility requirements?",
        "What is the examination pattern?",
        "Summarize the syllabus.",
        "What are the important rules?"
    ]
)

demo.launch(
    share=True
)
```
