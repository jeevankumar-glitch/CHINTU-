```python
import os
import ast
import gradio as gr

from typing import TypedDict
from getpass import getpass
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI

os.environ["GOOGLE_API_KEY"] = getpass("Enter your Gemini API Key: ")

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.2
)
```

```python
class DebugState(TypedDict):
    code: str
    error: str
    diagnosis: str
    fix: str
    validation: str
    report: str
```

```python
def analyze(state):
    code = state["code"]
    error = state["error"]

    prompt = f"""
You are a senior software debugging engineer.

Analyze this code and error.

CODE:
{code}

ERROR:
{error}

Identify:
1. Programming language
2. Error type
3. Root cause
4. Exact problematic part

Give a concise technical diagnosis.
"""

    response = llm.invoke(prompt)

    return {"diagnosis": response.content}
```

```python
def generate_fix(state):
    prompt = f"""
You are an expert programmer.

CODE:
{state["code"]}

ERROR:
{state["error"]}

DIAGNOSIS:
{state["diagnosis"]}

Generate a corrected version of the code.

Also explain exactly what you changed.

Return:
CORRECTED CODE:
<code>

EXPLANATION:
<explanation>
"""

    response = llm.invoke(prompt)

    return {"fix": response.content}
```

```python
def validate(state):
    code = state["code"]

    try:
        ast.parse(code)
        result = "✓ Original Python code has valid syntax."
    except SyntaxError as e:
        result = f"✗ Python syntax error detected: {e}"

    return {"validation": result}
```

```python
def report(state):
    prompt = f"""
Create a clear software debugging report.

DIAGNOSIS:
{state["diagnosis"]}

PROPOSED FIX:
{state["fix"]}

VALIDATION:
{state["validation"]}

Format:

🐛 ERROR ANALYSIS
...

🔍 ROOT CAUSE
...

🛠️ PROPOSED FIX
...

✅ VALIDATION
...

Keep it concise and easy for a student to understand.
"""

    response = llm.invoke(prompt)

    return {"report": response.content}
```

```python
graph = StateGraph(DebugState)

graph.add_node("analyze", analyze)
graph.add_node("fix", generate_fix)
graph.add_node("validate", validate)
graph.add_node("report", report)

graph.add_edge(START, "analyze")
graph.add_edge("analyze", "fix")
graph.add_edge("fix", "validate")
graph.add_edge("validate", "report")
graph.add_edge("report", END)

debugger = graph.compile()
```

```python
def debug_code(code, error):
    result = debugger.invoke({
        "code": code,
        "error": error
    })

    return result["report"]
```

```python
demo = gr.Interface(
    fn=debug_code,
    inputs=[
        gr.Code(
            label="💻 Enter Your Code",
            language="python"
        ),
        gr.Textbox(
            label="🐛 Error Message",
            placeholder="Paste the error message here..."
        )
    ],
    outputs=gr.Markdown(label="🤖 AI Debugging Report"),
    title="🧑‍💻 AI Software Debugging Agent",
    description="Powered by Gemini + LangGraph"
)

demo.launch()
```
