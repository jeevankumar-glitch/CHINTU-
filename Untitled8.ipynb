````python
import crewai

print("CrewAI:", crewai.__version__)

from crewai import Agent, Task, Crew

print("✅ CrewAI is working!")
````

````python
import os
from getpass import getpass

GEMINI_API_KEY = getpass("Enter your Gemini API key: ")

os.environ["GEMINI_API_KEY"] = GEMINI_API_KEY

print("✅ API key loaded")
````

````python
from crewai import LLM

llm = LLM(
    model="gemini/gemini-3.6-flash",
    api_key=GEMINI_API_KEY
)

print("✅ Gemini connected to CrewAI")
````

````python
from crewai import Agent

syllabus_agent = Agent(
    role="Syllabus Analyzer",
    goal="Break the student's syllabus into clear subjects and topics.",
    backstory="You are an expert academic curriculum analyst.",
    llm=llm,
    verbose=True
)

priority_agent = Agent(
    role="Priority Analyst",
    goal="Identify which study topics should receive the most attention.",
    backstory="You are an expert academic planner.",
    llm=llm,
    verbose=True
)

planner_agent = Agent(
    role="Study Planner",
    goal="Create a realistic day-by-day study timetable.",
    backstory="You are an expert study schedule designer.",
    llm=llm,
    verbose=True
)

coach_agent = Agent(
    role="Study Coach",
    goal="Review and improve the final study timetable.",
    backstory="You are an experienced academic coach.",
    llm=llm,
    verbose=True
)

print("✅ Four agents created")
````

````python
student_name = input("Student name: ")

subjects = input(
    "Subjects separated by commas: "
)

exam_date = input(
    "Exam date: "
)

hours = input(
    "Available study hours per day: "
)

level = input(
    "Preparation level (Beginner/Intermediate/Advanced): "
)

syllabus = input(
    "Enter your syllabus/topics: "
)
````

````python
from crewai import Task

task1 = Task(
    description=f"""
    Analyze this student's syllabus and **return the output in a JSON object as specified in the `expected_output`**.

    Student: {student_name}

    Subjects: {subjects}

    Exam date: {exam_date}

    Preparation level: {level}

    Syllabus:
    {syllabus}

    Identify every important topic and estimate its
    difficulty and study requirement.
    """,

    expected_output="""
    Return a JSON object where keys are subjects and values are arrays of
    topics. Each topic should be an object with 'topic', 'difficulty' (e.g., 'Low', 'Medium', 'High'),
    and 'study_requirement' (e.g., 'Low', 'Medium', 'High').

    Example:
    ```json
    {
        "DBMS": [
            {"topic": "SQL", "difficulty": "Medium", "study_requirement": "High"},
            {"topic": "Normalization", "difficulty": "High", "study_requirement": "High"}
        ],
        "Java": [
            {"topic": "OOP", "difficulty": "Medium", "study_requirement": "High"}
        ]
    }
    ```
    """,

    agent=syllabus_agent
)


task2 = Task(
    description="""
    Using the syllabus analysis from the previous agent,
    classify the topics into:

    HIGH PRIORITY
    MEDIUM PRIORITY
    LOW PRIORITY

    Consider difficulty, importance, preparation level
    and examination date.
    """,

    expected_output="""
    A prioritized list of all major topics.
    """,

    agent=priority_agent
)


task3 = Task(
    description=f"""
    Create a day-by-day study timetable.

    Student: {student_name}

    Exam date: {exam_date}

    Available study time: {hours} hours per day.

    Preparation level: {level}

    The timetable must contain:

    - Subject
    - Topic
    - Duration
    - Breaks
    - Revision

    Do not overload the student.
    """,

    expected_output="""
    A complete personalized study timetable.
    """,

    agent=planner_agent
)


task4 = Task(
    description="""
    Review the timetable created by the previous agent.

    Check whether:

    - The workload is realistic.
    - Important topics have enough time.
    - Revision is included.
    - Breaks are included.
    - Subjects are reasonably distributed.

    Fix problems and produce the final study plan.
    """,

    expected_output="""
    A polished final study plan with daily schedule,
    revision strategy and study recommendations.
    """,

    agent=coach_agent
)
````

````python
from crewai import Crew, Process

crew = Crew(
    agents=[
        syllabus_agent,
        priority_agent,
        planner_agent,
        coach_agent
    ],

    tasks=[
        task1,
        task2,
        task3,
        task4
    ],

    process=Process.sequential,

    verbose=True
)

print("✅ StudyPilot Crew ready!")
````

````python
result = await crew.kickoff_async()
print("\n")
print("=" * 70)
print("🧠 STUDYPILOT AI — YOUR FINAL STUDY PLAN")
print("=" * 70)
print()

print(result)
````
